# 10.8 Python Errors

## Syntax Error

A compile-time (Translation) error that breaks language rules and is caught before the program runs, thus stopping the program from being able to run.

## Exception (Run-Time) Error

The syntax is fine (it is legal) so the program runs, however, because of this Exception, the program will stop and give you an error.

## Logic Error

Type of run-time error. The program runs normally and gives you output as it is supposed to. The problem is, if you aren't paying attention, the output is INCORRECT.

Logic Errors can be difficult to spot. This is why you have to have an understanding of what the output should be.

# 10.1 Algorthm Design

## Algorithm

A recipe for a solution.

## Algorithms should be:

### Unambiguous

Precise instructions at every step. No guesswork or personal opinion.

### Executable

Every step can actually be carried out in practice.

### Terminating

The steps eventually come to an end.

### How do we describe an algorithm in pseudocode?

1. Inputs & Outputs - What goes in? What do we want to come out?  
2. Break it down - split the larger problem into smaller, executable subtasks.  
3. Write pseudocode - Order the steps so values exist before they're used.  
4. Test it - Work a problem by hand with sample values.
